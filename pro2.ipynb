```python
import tensorflow as tf
import numpy as np
%pip install matplotlib

import matplotlib.pyplot as plt
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28,28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
training = model.fit(x_train, y_train, epochs=5, verbose=1)
```

```python
from PIL import Image

filename = "number6.webp"
image = Image.open(filename)
```

```python
image = image.convert("L")
gray = np.array(image)

# Convert a light-background image to MNIST's dark-background format.
if gray.mean() > 127:
    gray = 255 - gray

foreground = gray > 30
rows, columns = np.where(foreground)
if len(rows) == 0:
    raise ValueError("No digit found in the input image")

cropped = gray[rows.min():rows.max() + 1, columns.min():columns.max() + 1]
cropped_image = Image.fromarray(cropped.astype(np.uint8))
cropped_image.thumbnail((20, 20), Image.Resampling.LANCZOS)

canvas = Image.new("L", (28, 28), 0)
offset = ((28 - cropped_image.width) // 2, (28 - cropped_image.height) // 2)
canvas.paste(cropped_image, offset)

image = np.array(canvas, dtype="float32") / 255.0
image = np.expand_dims(image, axis=0)
```

```python
image = np.asarray(image, dtype="float32").reshape(1, 28, 28)
prediction = model.predict(image, verbose=0)
predicted_digit = int(np.argmax(prediction[0]))
confidence = float(np.max(prediction[0]))
print(f"The predicted digit for the input image is: {predicted_digit}")
print(f"Confidence: {confidence:.2%}")
```
