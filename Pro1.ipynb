```python
exp = np.array([1, 2, 3, 4, 5, 6], dtype = float)

train = np.array([20, 30, 40, 50, 60, 65], dtype = int)

projects = np.array([2, 3, 4, 5, 6, 7], dtype = int)

dept = np.array([ "IT", "HR", "IT", "Finance", "IT", "HR" ], dtype = str)

perf = np.array([45, 52, 61, 68, 75, 78], dtype = int)
```

```python
encoder = OneHotEncoder()
department_encode = encoder.fit_transform(dept.reshape(-1, 1))
print(department_encode)
```

```python
X = np.column_stack((exp, train, projects, department_encode.toarray(),perf))
Y = perf
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X.shape[1],)),
    tf.keras.layers.Dense(units=10, activation = 'relu', input_shape=(X.shape[1],)),
    tf.keras.layers.Dense(1)
])
```

```python
model.compile(
    optimizer = 'adam',
    loss = 'mean_squared_error'
)
```

```python
training = model.fit(X, Y, epochs = 20, verbose = 1)
```
