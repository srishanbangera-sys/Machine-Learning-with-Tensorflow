```python
exp = np.array([1, 2, 3, 4, 5, 6], dtype = float)

train = np.array([20, 30, 40, 50, 60, 65], dtype = int)

projects = np.array([2, 3, 4, 5, 6, 7], dtype = int)

dept = np.array([ "IT", "HR", "IT", "Finance", "IT", "HR" ], dtype = str)

perf = np.array([45, 52, 61, 68, 75, 78], dtype = int)
```

```python
encoder = OneHotEncoder()
department_encode = encoder.fit_transform(dept.reshape(-1, 1))
print(department_encode)
```

```python
X = np.column_stack((exp, train, projects, department_encode.toarray()))
Y = perf
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X.shape[1],)),
    tf.keras.layers.Dense(4, activation='relu'),
    tf.keras.layers.Dense(1)
])
```

```python
model.compile(
    optimizer = 'sgd',
    loss = 'mean_squared_error'
)
```

```python
training = model.fit(X, Y, epochs = 20, verbose = 1)
```

```python
test_exp = np.array([7], dtype = float)
test_train = np.array([70], dtype = int)
test_projects = np.array([8], dtype = int)

prediction = model.predict(np.column_stack((test_exp, test_train, test_projects, encoder.transform(np.array(["IT"]).reshape(-1, 1)).toarray())))\

for i in range(len(test_exp)):
    print(f"Predicted performance for experience {test_exp[i]}: {prediction[i][0]}")
```
