```python
import tensorflow as tf
```

```python
scalar = tf.constant(20)
print(scalar)
```

```python
vector = tf.constant([10, 20, 30, 40])
print(vector)
```

```python
matrix = tf.constant([[1, 2, 3], [4, 5, 6]])
print(matrix)

print(tf.cast(matrix, tf.float32))
```

```python
ones = tf.ones((2, 3))
print(ones)
```

```python
zeros = tf.zeros((2, 3))
print(zeros)
```

```python
range = tf.range(1, 11)
print(range)

range = tf.cast(range, tf.float32)
print(range)

print(tf.rank(range))
print(range.dtype)
```

```python
matrix = tf.constant([[30, 40, 50], [60, 70, 80], [90, 100, 110]])

matrix = tf.transpose(matrix)
print(matrix)
```

```python
a = tf.constant([[10, 20, 30]])
b = tf.constant([[1], [2], [3]])

sum = tf.add(a, b)
print(sum)

diff = tf.subtract(a, b)
print(diff)

multi = tf.multiply(a, b)
print(multi)

div = tf.divide(a, b)
print(div)

square = tf.square(a)
print(square)

root = tf.sqrt(tf.cast(a, tf.float32))
print(root)


```

```python
overallsum = tf.reduce_sum(a)
print(overallsum)   

mean = tf.reduce_mean(a)
print(mean)

mode = tf.reduce_max(a)
print(mode)

min = tf.reduce_min(a)
print(min)

max = tf.argmax(a)
print(max)
```
